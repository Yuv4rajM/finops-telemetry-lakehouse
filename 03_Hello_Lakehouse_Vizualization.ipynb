```python
%sql
SELECT 
date_trunc('month', usage_date) as billing_month,
description as cloud_service,
round(sum(Total_Net_Cost), 4) monthly_cost
FROM workspace.gold.gcp_billing_final
GROUP BY billing_month, cloud_service
ORDER BY billing_month DESC, monthly_cost DESC
```

```python
%sql
SELECT 
  sku.description AS specific_charge,
  ROUND(SUM(net_cost), 4) AS total_cost
FROM workspace.silver.gcp_billing_clean
WHERE service.description = 'Compute Engine'
GROUP BY specific_charge
ORDER BY total_cost DESC;
```
