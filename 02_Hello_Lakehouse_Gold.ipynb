#Init

```python
from pyspark.sql.functions import sum as spark_sum, round as spark_round
```

#Reading from Silver Table

```python
silver_df = spark.read.table("workspace.silver.gcp_billing_clean")
```

#Business Logic

```python
#group by the date and the specific cloud service, then sum the net cost
gold_df = silver_df.groupBy("usage_date", "service.description").agg(spark_round(spark_sum("net_cost"), 4).alias("Total_Net_Cost"))
```

```python
display(gold_df)
```

##Writing Data into Gold Layer

```python
gold_df.write.mode("overwrite").saveAsTable("workspace.gold.gcp_billing_final")
print("Data landed successfully in gold layer")
```
